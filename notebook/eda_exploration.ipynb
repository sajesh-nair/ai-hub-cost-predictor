```python
import pandas as pd
import numpy as np
import os

# Set file path to our real dataset
csv_path = os.path.join("llm_price_performance_tracker_2026-03-31.csv")

# Load data
df = pd.read_csv(csv_path)
```

```python
# 1. Check the dimensions
print(f"Dataset Shape: {df.shape}")
```

```python
# 2. View the head
df.head()
```

```python
# 1. Inspect data types and structural integrity
print("--- DataFrame Information ---")
df.info()
```

```python
print("\n--- Missing Values Count per Column ---")
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

```python
# Check the statistical spread of cost, throughput, and parameter metrics
target_metrics = [
    'input_cost_usd_per_1m', 
    'output_cost_usd_per_1m', 
    'output_tokens_per_second', 
    'time_to_first_token_s',
    'parameter_count'
]
```

```python
# Ensure we only look at columns that exist in your slice
valid_metrics = [col for col in target_metrics if col in df.columns]
df[valid_metrics].describe()
```

```python
df['provider'].value_counts()
```

```python
# Check the spread and median values of our primary cost target
df['blended_cost_usd_per_1m'].describe()
```

```python
# Calculate the exact numerical skew of the cost target
df['blended_cost_usd_per_1m'].skew()
```

```python
# Apply log transformation and check the new skewness
log_skew = np.log1p(df['blended_cost_usd_per_1m']).skew()
print(f"Original Skewness: {df['blended_cost_usd_per_1m'].skew():.4f}")
print(f"Log-Transformed Skewness: {log_skew:.4f}")
```

```python
# Check the statistical spread of our core composite intelligence metric
df['composite_benchmark'].describe()
```

```python
# Show rows where the composite benchmark score is missing
df[df['composite_benchmark'].isnull()][['model_name', 'provider']]
```

```python
# Calculate the correlation between the absolute intelligence index and our composite score
df[['aa_intelligence_index', 'composite_benchmark']].corr()
```

```python
# Group by provider and look at the average cost and intelligence score
df.groupby('provider')[['blended_cost_usd_per_1m', 'composite_benchmark']].mean().sort_values(by='composite_benchmark', ascending=False).head(10)
```

```python
# Compare average cost based on whether a model is open source or proprietary
df.groupby('is_open_source')['blended_cost_usd_per_1m'].mean()
```
